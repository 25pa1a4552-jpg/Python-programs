```python
# Program that counts the number of vowels in a string
 text=input("Enter a string:")
 vowels="aeiouAEIOU"
 count=0
 for char in text:
  if char in vowels:
    count+=1
    print("Number of vowels:", count)
```

```python
# code that prints reverse of the numbers
def reverse_number(n):
  rev=0
  while n>0:
    digit=n%10
    rev=rev*10+digit
    n=n//10
    return rev
    num=int(input("Enter a number:"))
    result=reverse_number(num)
    print("Reverse of the number:",result)
```
