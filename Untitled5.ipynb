```python
#program that reverse a string using function
def reverse(string):
  s = string[::-1]
  return s
reverse("python")
```

```python
#Program that reverse a string using user-defined function
def rev(string):
  rev= ""
  for i in string:
    rev = i + rev
  return rev
rev("PYTHONPROGRAMING")
```

```python
#Function that return multiple values
def multiple_function(*num):
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
res_sum,res_count,res_avg=multiple_function(1,2,3,4,5,6,8)
print(f"the sum of number is {res_sum}, the count is {res_count},the average is {res_avg:.2f}")
```
